## 1. Method choice and why

*Which method from the toolkit, and why it fits your lane.*

We evaluate Logistic Regression, Random Forest, and XGBoost (Ranking / Tree-based) against our Week 4 heuristic baseline:

Logistic Regression: Serves as an interpretable pointwise linear baseline to determine whether learned linear weights over features like CTR, rank position, and staleness provide measurable lift over hand-crafted scoring rules.

Random Forest: Captures non-linear feature interactions (such as the steep degradation of clicks past top positions combined with high staleness) without requiring manual binning, while offering reliable permutation-based feature attribution.

XGBoost: Optimizes directly on non-linear gradient boosting paths, handling tabular scale disparities and mixed discrete/continuous signal distributions effectively.

## 2. Split design

*Grouped by client? Time-aware? Say why this split is honest for your question.*

Split Strategy: Grouped split by query / client entity (GroupShuffleSplit with a 60 / 20 / 20 train / validation / test division).

Honesty & Leakage Prevention: Standard random or stratified row splits cause severe data leakage by scattering candidate documents from the same query across train and test sets, artificially inflating ranking scores. Enforcing a strict group boundary ensures all candidate items for an evaluated query are entirely unseen during training, measuring true generalization.

## 4. Errors and interpretation

*Where is the model wrong? What does it lean on? A short error analysis beats a big metric table.*

Primary Signal Reliance: Permutation importance indicates the models lean most heavily on historical click-through rate and term overlap signals, with staleness acting as a secondary dampener.

Low-Volume / Tail Queries (False Negatives): Errors concentrate on long-tail queries with low impression counts (sparse historical CTR). The model struggles to surface relevant documents without sufficient engagement history.

Semantic Paraphrasing (False Positives/Negatives): Documents with high semantic intent match but low literal lexical overlap are systematically underscored, whereas generic high-impression documents occasionally receive overconfident rankings.
